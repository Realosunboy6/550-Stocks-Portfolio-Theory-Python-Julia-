Pull any slice of the built-in universe (11 GICS sectors ≈ 400 curated stocks + 300 ETFs in 22 categories), cache to Google Drive so phone reruns are instant, and eyeball the data.

```python
#@title Setup — run this first {display-mode: "form"}
try:
    import portlab
except ImportError:
    %pip install -q "portlab @ git+https://github.com/Realosunboy6/550-Stocks-Portfolio-Theory-Python-Julia-.git"
    import portlab
print("portlab", portlab.__version__, "ready")
```

```python
#@title Optional: mount Google Drive for a persistent cache {display-mode: "form"}
use_drive_cache = False  #@param {type:"boolean"}
if use_drive_cache:
    from portlab.data import mount_drive
    mount_drive()
from portlab.data import cache_root
print("Cache directory:", cache_root())
```

```python
#@title Universe browser {display-mode: "form"}
from portlab.data import ALL_TICKERS, ETF_CATEGORIES, SECTORS, ASSET_CLASS_PROXIES
print(f"{len(ALL_TICKERS)} tickers total")
for s, ts in SECTORS.items():
    print(f"  {s:26s} {len(ts):4d} stocks")
for c, ts in ETF_CATEGORIES.items():
    print(f"  {c:26s} {len(ts):4d} ETFs")
print("\nAsset-class proxies (PV-style):")
for name, (tick, yr) in ASSET_CLASS_PROXIES.items():
    print(f"  {name:28s} {tick:8s} since ~{yr}")
```

```python
#@title Download a slice {display-mode: "form"}
sector = "Information Technology"  #@param ["Communication Services", "Consumer Discretionary", "Consumer Staples", "Energy", "Financials", "Health Care", "Industrials", "Information Technology", "Materials", "Real Estate", "Utilities"]
start_date = "2021-01-01"  #@param {type:"date"}
SMOKE = False
```

```python
from portlab.data import get_prices, sector_tickers
from portlab.returns import clean_returns, simple_returns
from portlab import plots

ticks = sector_tickers(sector)[: (5 if SMOKE else None)]
prices = get_prices(ticks, start_date)
rets = clean_returns(simple_returns(prices))
print(f"{rets.shape[1]} tickers x {rets.shape[0]} days (cached for instant reruns)")
plots.growth_chart(rets.mean(axis=1).rename(f"{sector} equal-weight"), initial=10000).show()
```
